## Q6. Median value of horsepower
1. Hallar la mediana de `horsepower`.
2. Calcular el valor más frecuente (moda) de `horsepower`.
3. Rellenar los valores nulos con la moda.
4. Recalcular la mediana de `horsepower`.

¿Ha cambiado? (`Yes, it increased`, `Yes, it decreased`, `No`)

```python
# Pregunta 6: Mediana antes y después de imputar moda
horsepower_col = df["horsepower"]
median_before = horsepower_col.median()
mode_hp = horsepower_col.mode().iloc[0]

horsepower_imputed = horsepower_col.fillna(mode_hp)
median_after = horsepower_imputed.median()

print(f"Mediana inicial: {median_before}")
print(f"Moda imputada:   {mode_hp}")
print(f"Mediana final:   {median_after}")

if median_after < median_before:
    res_q6 = "Yes, it decreased"
elif median_after > median_before:
    res_q6 = "Yes, it increased"
else:
    res_q6 = "No"

print(f"Resultado: {res_q6}")
```

## Q7. Sum of weights (Linear Regression)
1. Filtrar autos de Asia.
2. Seleccionar columnas `vehicle_weight` y `model_year`.
3. Tomar los primeros 7 valores (`X`).
4. Calcular $X^T X$.
5. Invertir $X^T X$.
6. Crear $y = [1100, 1300, 800, 900, 1000, 1100, 1200]$.
7. Calcular $w = (X^T X)^{-1} X^T y$.
8. ¿Cuál es la suma de los elementos de $w$?

```python
# Pregunta 7: Implementación normal de regresión lineal
asia_subset = df.loc[df["origin"] == "Asia", ["vehicle_weight", "model_year"]].head(7)
X = asia_subset.to_numpy()
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
w = XTX_inv @ X.T @ y

print(f"Matriz X (7x2):\n{X}")
print(f"Vector y: {y}")
print(f"Pesos w: {w}")
print(f"Suma de elementos de w: {w.sum():.3f}")
```
