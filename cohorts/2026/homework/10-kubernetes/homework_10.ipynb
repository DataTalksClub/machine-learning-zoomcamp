### Question 4: Service type

El servicio predeterminado del clúster `kubernetes` y el servicio definido en `service.yaml` utilizan el tipo predeterminado de red interna de Kubernetes: **`ClusterIP`**.

````python
from pathlib import Path
import yaml

manifest_dir = Path(".").resolve()
if not (manifest_dir / "service.yaml").exists():
    manifest_dir = Path("cohorts/2026/homework/10-kubernetes").resolve()

service_yaml = yaml.safe_load((manifest_dir / "service.yaml").read_text(encoding="utf-8"))
service_type = service_yaml["spec"]["type"]

print(f"Tipo de servicio declarado en service.yaml: '{service_type}'")
print(f"Respuesta Pregunta 4: {service_type}")
````

### Question 6: Container port

Inspeccionamos `deployment.yaml` para comprobar el puerto expuesto por el contenedor:

````python
deployment_yaml = yaml.safe_load((manifest_dir / "deployment.yaml").read_text(encoding="utf-8"))
container_port = deployment_yaml["spec"]["template"]["spec"]["containers"][0]["ports"][0]["containerPort"]

print(f"Puerto de contenedor declarado en deployment.yaml: {container_port}")
print(f"Respuesta Pregunta 6: {container_port}")
````
