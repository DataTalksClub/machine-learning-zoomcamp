```python

import subprocess
import json

result = subprocess.run(
    ["python", "q6_test.py"],
    capture_output=True,
    text=True
)

print(result.stdout)

if result.returncode != 0:
    print(result.stderr)
else:
    try:
        prediction = json.loads(result.stdout)
        probability = prediction["conversion_probability"]
        print(f"Question 1 answer: {probability:.3f}")
    except (json.JSONDecodeError, KeyError, TypeError):
        print("Read conversion_probability from the output above.")
```
